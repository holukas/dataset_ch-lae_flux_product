# Prepare data for REddyProc USTAR threshold detection

```python
from datetime import datetime
from pathlib import Path

import diive as dv
from diive.core.io.files import save_parquet, load_parquet
from diive.core.times.times import insert_timestamp
```

## Load data

```python
SOURCEDIR = r"."
FILENAME = r"12_IRGA72_FluxProcessingChain_after-L3.2_NEE_QCF11_2016-2024.parquet"
FILEPATH = Path(SOURCEDIR) / FILENAME
print(f"Data will be loaded from the following file:\n{FILEPATH}")
maindf = load_parquet(filepath=FILEPATH)
maindf
```

```python
df = insert_timestamp(data=maindf, convention="end", insert_as_first_col=True, set_as_index=True)
df
```

## Subset

```python
[print(c) for c in df.columns if "SIGMA" in c]
```

```python
subsetcols = ["NEE_L3.1_L3.2_QCF", "USTAR", "SW_IN_T1_47_1_gfXG", "TA_T1_47_1_gfXG", "W_SIGMA"]
subset = df[subsetcols].copy()
# subset = subset.dropna()
subset
```

## Stats

```python
display(subset.dropna())
```

## Save data

```python
filename = "14_IRGA72_SUBSET_NEE_QCF11_2016-2024"
subset.to_csv(f"{filename}.csv")
save_parquet(data=subset, filename=filename)
```
