## ℹ️ About this notebook
Download raw soil temperature from the InfluxDB database, remove the worst outliers on the **high-resolution** data, resample to 30MIN, and upload the result back to the database. Screening uses `StepwiseMeteoScreeningDb` from [diive](https://github.com/holukas/diive) (`diive/preprocessing/qaqc/meteoscreening.py`); download and upload use diive's in-house InfluxDB engine (`InfluxIO`, in `diive/core/io/db/influx`).

**Flow:** download (`InfluxIO`) → screen on high-res data (`diive`) → resample to 30MIN → upload.

**Scope is deliberately narrow.** The raw record is millions of rows, so anything done here is expensive. This notebook only takes out what is unambiguously wrong at high resolution — the 2022 fault window (manual removal) — and hands over a clean 30MIN series. Everything else (sensor comparisons, level/offset questions, gap handling, homogenisation across sensor generations) is settled later in `30_PRODUCTS/`, **on the half-hourly data**, where it costs a fraction as much.

**Outlier detection is stepwise:** run a test, inspect its preview plot, then commit it with `mscr.addflag()`. Re-run with different parameters as often as you like before committing. Run only the tests a variable actually needs. At the end all committed flags are aggregated into one overall quality flag `QCF`.

> **Which sensor this is.** In this profile the replicate suffix names the sensor family, not the depth: `_1` = the water-potential sensors of the 19 Mar 2020 install (MPS-2 at 0.05/0.1/0.6 m, TEROS 21 at 0.2/0.3/0.5 m), `_2` = the TEROS 12 probes that also report `SWC_FF1_<depth>_1`, `_3`/`_4` = Campbell 109 thermistors added at the March 2021 logger rebuild. **This one is a Campbell 109 thermistor (analogue, not on the SDI-12 bus).** Being analogue rather than SDI-12 is what makes this notebook differ from its `_1`/`_2` siblings, in three places marked below.

> **Database access** needs the `influxdb-client` package, which this project pulls in via the **`diive[db]` extra** (declared in `pyproject.toml`). diive *also* ships a `db` dependency group, but dependency groups are local to the project that declares them — `uv sync --group db` only works inside the diive repo, not from here.

## ⏱️ Timestamp convention (important)
The database stores every timestamp in **UTC** and as **`TIMESTAMP_END`** (the stamp marks the *end* of the averaging interval). Getting this right is the one thing that must not go wrong: the value written back to the database depends on it, and so does every window in `REMOVE_DATES`.

The single knob is `TIMEZONE_OFFSET_TO_UTC_HOURS` (set in *User settings*). It is applied **identically** on download and on upload:

| Stage | Timezone | Convention | Done by |
|---|---|---|---|
| Database | UTC | `TIMESTAMP_END` | InfluxDB |
| After `dbc.download(..., timezone_offset_to_utc_hours=N)` | local (UTC+N) | `TIMESTAMP_END` | InfluxIO |
| During screening | local | `TIMESTAMP_MIDDLE` (converted internally) | `StepwiseMeteoScreeningDb` |
| After `mscr.resample()` | local | back to `TIMESTAMP_END` | diive |
| After `dbc.upload_singlevar(..., timezone_offset_to_utc_hours=N)` | UTC | `TIMESTAMP_END` | InfluxIO |

**One place where this bites.** `REMOVE_DATES` is matched against `TIMESTAMP_MID`, not `TIMESTAMP_END`: `StepwiseMeteoScreeningDb` hands the series to `ManualRemoval` *after* the internal conversion, so the record whose `TIMESTAMP_END` is `10:39:00` carries the label `10:38:30` inside the test. A bare `'2025-12-19 10:39:00'` would therefore match **nothing** and the removal would silently do nothing. It only matters when removing individual records: a window spanning days has edges far larger than the half-record offset.

## 🔍 Outlier detection
Run a test → inspect its preview → commit with `mscr.addflag()`. Only the committed flag of the **most recent** test is added. Skip any test a variable does not need.

**What is committed here**: the **2022 fault window** (manual removal), a physical-range backstop (absolute limits), and the **missing-values** flag. Why the rest are off:

- **Never split day and night.** Soil temperature does have a diurnal cycle, but at this depth it is damped and lagged relative to the surface, so a solar-geometry split does not line up with it — it would only halve the sample behind every threshold. Every test below runs with `separate_daytime_nighttime=False`.
- **Distribution-wide tests measure the season, not a fault.** A z-score over the whole record asks how far a value sits from the multi-year mean; January at 2 °C and July at 19 °C are both correct.
- **Differencing-based tests would actually work here — and are still not committed.** This is the one place where this notebook differs from its `_1`/`_2` siblings on a point of arithmetic rather than taste. Those channels are SDI-12 and quantised to a 0.05 °C grid, which collapses the rolling MAD a Hampel filter scales its band by to exactly zero. This one is an **analogue** thermistor: its values are continuous (only 0.007 % land on that grid, i.e. at chance) and it has a **single** 1MIN resolution era with no back-filling. So a spike test has a real scale to work with. It is left off because there is nothing for it to find, not because it would misbehave — see *Other tests* below. If any channel in this family ever justifies one, it is this one.

### Manual removal
Flag specific timestamps or time ranges for removal - known sensor failures, maintenance windows, logger artefacts. Give `[start, stop]` pairs and/or single timestamps.

**This is the one channel at 0.05 m that needs a substantial window**, and it is also the only thing in this notebook that deletes data on a person's say-so — so the cell below **re-derives its own justification from the downloaded record on every run** and refuses to proceed if the evidence has changed.

Between **2022-02-24 and 2022-05-12** this thermistor swings ±25 K a day. Soil at 0.05 m under a mixed forest cannot do that; the litter layer alone damps the surface signal far below it, and this channel's own twin `TS_FF1_0.05_3` — the same make of sensor, the same nominal depth, a few metres away — sits at 1.8-9 °C with a 0.5-1.3 K amplitude through exactly the same weeks. The reading is not soil temperature. The most likely explanation is that the probe was out of the ground, but the fieldbook records nothing for the period, so the removal rests on physical evidence alone.

> ⚠️ **A second, quieter episode is *not* removed here.** After the 23 Oct 2025 windstorm this channel shows a median daily amplitude of 3.38 K in Nov 2025 and 1.86 K in Dec 2025, against 0.37-0.71 K in every earlier Nov/Dec. That is elevated but an order of magnitude below the 2022 fault, and the storm is documented as damaging *above-ground* sensors. It is left in the record and flagged for `30_PRODUCTS/` to settle on the 30MIN data, where the comparison against `0.05_3` is cheap. Removing it here on a hunch would be exactly the hand-removal-in-disguise this folder warns about.

> **On the open fieldbook question.** A 2026-07-17 `CHECK` entry asks whether this sensor is "too close to the surface? Very strong daily cycle". Outside the two episodes above the record does **not** support that: this channel runs at 1.0-1.7 K summer amplitude against 0.9-1.2 K for `0.05_3`. The strong cycle is episodic, not permanent.

```python
if SHOW_PARAM_HELP:
    help(dv.outliers.ManualRemoval)
```

```python
# --- 2022-02-24 -> 2022-05-12: this thermistor was not in the soil -----------------------
# Evidence, recomputed from this channel's own record by the cell below on every run:
#   * 67 days in the window carry a daily amplitude above 5 K;
#   * median daily amplitude 22.36 K in March 2022, against 1.30 K in March 2023;
#   * the window is where this channel's -5.49 and +31.53 degC extremes come from;
#   * its twin TS_FF1_0.05_3, a 109 thermistor 5 cm deep a few metres away, stays between
#     1.8 and 9 degC with 0.5-1.3 K amplitude through the same weeks.
# Soil at 0.05 m under a forest cannot swing 22 K in a day. A sensor lying on the litter can.
# The fieldbook has no entry for it; the removal rests on physical evidence alone, which is
# the same basis on which the Aug-Oct 2020 SWC decoupling was removed.
#
# (!) Boundaries carry a day of slack: they come from a daily statistic, so they are dates,
#     not times. This is a long window, so it is NOT the one-minute bracket the _1/_2
#     notebooks use - the TIMESTAMP_MID subtlety only matters for single-record removals.
REMOVE_DATES = [
    ['2022-02-24 00:00:00', '2022-05-12 23:59:00'],
]

# --- The window must defend itself, every run ---------------------------------------------
# Nothing here is quoted from the prose above; it is recomputed from the downloaded record.
_h = data_detailed[FIELD][FIELD].dropna()
_amp = _h.resample('D').max() - _h.resample('D').min()
_w0, _w1 = pd.Timestamp(REMOVE_DATES[0][0]), pd.Timestamp(REMOVE_DATES[0][1])
_inside = _amp.loc[_w0:_w1].dropna()
_outside = _amp.drop(_amp.loc[_w0:_w1].index).dropna()
print(f'daily amplitude inside the window:  median {_inside.median():.2f} K, '
      f'max {_inside.max():.2f} K, {int((_inside > 5).sum())} of {len(_inside)} days above 5 K')
print(f'daily amplitude outside it:         median {_outside.median():.2f} K, '
      f'{int((_outside > 5).sum())} of {len(_outside)} days above 5 K')
assert _inside.median() > 4 * _outside.median(), (
    '(!) the window no longer stands out from the rest of the record - re-derive it')
assert (_inside > 5).sum() >= 50, (
    '(!) fewer than 50 high-amplitude days inside the window - re-derive it')
print('PASSED - the window re-derived itself from this run\'s data.')

if REMOVE_DATES:
    mscr.flag_manualremoval_test(remove_dates=REMOVE_DATES, showplot=True, verbose=True)
else:
    print('No manual removal windows for this sensor - skip the addflag cell below.')
```

## ⬆️ Upload data to database

**Re-uploading overwrites the same variant - safe to re-run.** With `delete_from_db_before_upload=True` (below), the upload first *deletes*, then writes. The delete is scoped to the exact match `_measurement` + `varname` + `data_version` (`meteoscreening_diive`) over the uploaded time range, so re-screening a period replaces only its previous screened result. It never touches the raw data (different `data_version`, and a different `_raw` bucket), other variables, or other data versions. The delete-first step (rather than a plain overwrite) matters because InfluxDB keys a point by its full tag set: if a tag changed between runs (e.g. `units`, `gain`, `offset`), a plain write would leave the old point as a **duplicate** - the delete removes it regardless of tags.

```python
print(f'Uploading to bucket {BUCKET_PROCESSED}')
for v in mscr.resampled_detailed.keys():
    dbc.upload_singlevar(
        to_bucket=BUCKET_PROCESSED,
        to_measurement=assigned_measurements[v],
        var_df=mscr.resampled_detailed[v],
        timezone_offset_to_utc_hours=TIMEZONE_OFFSET_TO_UTC_HOURS,
        delete_from_db_before_upload=True,
    )
```
