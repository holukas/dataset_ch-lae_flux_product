# Merge IRGA72 fluxes + meteo 7 (2016-2024)

# Imports

```python
import importlib.metadata
import pandas as pd
import warnings
from datetime import datetime
from pathlib import Path

from diive.core.io.filereader import search_files, MultiDataFileReader
from diive.core.io.files import save_parquet, load_parquet
# pd.options.display.width = None
# pd.options.display.max_columns = None
pd.set_option('display.max_rows', 3000)
pd.set_option('display.max_columns', 3000)
warnings.filterwarnings(action='ignore', category=FutureWarning)
warnings.filterwarnings(action='ignore', category=UserWarning)
dt_string = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
version_diive = importlib.metadata.version("diive")
print(f"diive version: v{version_diive}")
```

# Load data: fluxes L1 and meteo (7 variables)

```python
irga72 = load_parquet(filepath="22.2_FLUXES_IRGA72_L1_2016-2024.parquet")
meteo7 = load_parquet(filepath=Path("../10_METEO/") / "12.5_METEO7_GAPFILLED_2004-2024.parquet")
```

# Merge files

## Start with IRGA data

```python
df = irga72.copy()
df
```

## Add *new* columns from meteo data

```python
newcols = [c for c in meteo7.columns if c not in df.columns]
df = pd.concat([df, meteo7[newcols]], axis=1)
display(newcols)
df
```

## Restrict data time range

```python
keep = (df.index.year >= 2016) & (df.index.year <= 2024)
df = df[keep].copy()
df
```

# Sort column order

```python
df = df.reindex(sorted(df.columns), axis=1)  # Sort column order
df
```

# Save to file

```python
OUTNAME = "22.4_FLUXES_L1_IRGA72+METEO7_2016-2024"
filepath = save_parquet(filename=OUTNAME, data=df)
# df.to_csv(f"{OUTNAME}.csv")
```
