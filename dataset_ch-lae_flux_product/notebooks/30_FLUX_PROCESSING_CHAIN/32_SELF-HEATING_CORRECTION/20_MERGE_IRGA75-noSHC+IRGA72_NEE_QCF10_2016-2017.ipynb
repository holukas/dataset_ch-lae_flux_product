# Merge IRGA75 (no self-heating correction) with IRGA72 data (2016-2017)

```python
from datetime import datetime
import pandas as pd
from diive.core.io.files import save_parquet, load_parquet
```

# Required variables

```python
# Variable names in EddyPro _fluxnet_ output files
AIR_CP = "AIR_CP"
AIR_DENSITY = "AIR_DENSITY"
VAPOR_DENSITY = "VAPOR_DENSITY"
U = "U"
USTAR = "USTAR"
TA = "TA_T1_47_1_gfXG"
SWIN = "SW_IN_T1_47_1_gfXG"
CO2_MOLAR_DENSITY = "CO2_MOLAR_DENSITY"
FLUX = "NEE_L3.1_L3.2_QCF"

usecols_irga75 = [FLUX, CO2_MOLAR_DENSITY]
usecols_irga72 = [FLUX, AIR_CP, AIR_DENSITY, VAPOR_DENSITY, U, USTAR, TA, SWIN, CO2_MOLAR_DENSITY]
```

# Load IRGA75 data

```python
df_irga75 = load_parquet(filepath="02_IRGA75_FluxProcessingChain_after-L3.2_NEE_QCF10_2016-2017.parquet")
# [print(c) for c in df_irga75 if "NEE" in c];
```

```python
df_irga75 = df_irga75[usecols_irga75].copy()
df_irga75 = df_irga75.add_suffix('_IRGA75')
df_irga75.describe()
```

# Load IRGA72 data

```python
df_irga72 = load_parquet(filepath="12_IRGA72_FluxProcessingChain_after-L3.2_NEE_QCF10_2016-2017.parquet")
```

```python
df_irga72 = df_irga72[usecols_irga72].copy()
df_irga72 = df_irga72.add_suffix('_IRGA72')
df_irga72.describe()
```

# Merge data

```python
df = pd.concat([df_irga72, df_irga75], axis=1)
df.describe()
```

# Save to file

```python
filename = "21_MERGED_IRGA75-noSHC+IRGA72_FluxProcessingChain_after-L3.2_NEE_QCF10_2016-2017"
df.to_csv(f"{filename}.csv", index=True)
save_parquet(data=df, filename=filename)
```
